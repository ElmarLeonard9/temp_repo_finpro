# **Section 1: Business Explanation**

## **1.1 Business Context**

Olist is a **Brazilian Software-as-a-Service (SaaS) technology company** that provides an ecosystem of solutions to help merchants manage and grow their businesses across e-commerce channels, including marketplace integration, order management, payments, and logistics.

The success of Olist can be reflected in **customer satisfaction**, which can be measured through the **review scores** provided by customers. Low customer satisfaction, indicated by low review scores, can negatively affect sellers' reputation and sales performance, which may ultimately impact the health of Olist's ecosystem. On the other hand, high customer satisfaction can support **customer retention and repeat purchases**, contributing to the long-term sustainability of the ecosystem.

Based on the Olist Public E-commerce Dataset (2016–2018), the **average review score is 4.09 out of 5**. Although **56.2% of orders received a 5-star rating**, **16.6% received a low rating (1–2 stars)**, indicating room for improvement in customer satisfaction.

The results of the preliminary analysis indicate a **relationship between delivery performance and customer satisfaction**. Among all orders in the dataset, **46.9% of late orders received a 1-star rating**, while **59.01% of on-time orders received a 5-star rating**. This pattern suggests that delivery performance may play an important role in shaping customer satisfaction.

Therefore, it's important to **identify the factors associated with late delivery** and understand the characteristics of orders that are at higher risk of experiencing delays. This understanding can help Olist evaluate its delivery performance and identify operational areas that require further attention to improve customer satisfaction.

## **1.2 Problem Statements**

Based on the background above, the main problems to be addressed are:

1. What factors are associated with late delivery, and what are the characteristics of orders that are at higher risk of experiencing delays?
2. How can a machine learning model be developed to predict the likelihood of late delivery, enabling Olist to take proactive actions to improve delivery performance and customer satisfaction?

## **1.3 Key Objectives**

To address the business problem, this project aims to identify factors associated with late delivery, predict high-risk orders, and provide actionable insights to improve delivery performance and customer satisfaction.

1. **Identify factors associated with late delivery**
   - Analyze order, seller, customer, product, and logistics-related factors that contribute to delivery delays.

2. **Understand high-risk order characteristics**
   - Identify patterns and characteristics of orders that are more likely to experience late delivery.

3. **Predict late delivery risk**
   - Develop a machine learning model to predict the likelihood of an order experiencing late delivery.

4. **Provide actionable insights**
   - Translate the analysis and model results into insights that support proactive actions for high-risk orders.

5. **Improve delivery performance and customer satisfaction**
   - Support Olist in reducing delivery delays and improving customer satisfaction.

**Key stakeholder: Operations & Logistics Team**

The findings can help them identify the factors behind delays, detect high-risk orders earlier, and prioritize proactive actions to improve delivery performance and customer satisfaction.

## **1.4 Analytical Approach**

The analytical approach for this project consists of several systematic stages:

| **Stage** | **Implementation** |
|---|---|
| `Data Understanding` | Understand the structure and characteristics of the dataset, while identifying potential data issues such as inconsistencies, missing values, and anomalies. |
| `Exploratory Data Analysis (EDA)` | Conduct descriptive analysis and data visualization to identify patterns, relationships between variables, and factors associated with late delivery. |
| `Data Preprocessing` | Prepare and clean the data by handling missing values, outliers, categorical variables, and other data quality issues before modeling. |
| `Feature Engineering` | Create relevant features and transform existing variables to better represent the factors that may contribute to late delivery and improve model performance. |
| `Modeling` | Experiment with several machine learning algorithms to develop a model that can predict the likelihood of an order experiencing late delivery. |
| `Model Evaluation` | Evaluate model performance using metrics that are appropriate for the prediction objective and aligned with the business problem. |
| `Business Recommendation` | Translate the EDA findings and model results into actionable insights that can help Olist identify high-risk orders and improve delivery performance. |
